```python
n = int(input("number of  list1 : "))
list1 = list(map(int, input("values of list1 : ").split()))

m = int(input("number of list2 : "))
list2 = list(map(int, input("values of list2 : ").split()))

i = 0
j = 0
carry = 0

result = []


while i < n or j < m or carry:

    digit1 = list1[i] if i < n else 0
    digit2 = list2[j] if j < m else 0

    total = digit1 + digit2 + carry

    result.append(total % 10)
    carry = total // 10

    i += 1
    j += 1
print(*result)
```
