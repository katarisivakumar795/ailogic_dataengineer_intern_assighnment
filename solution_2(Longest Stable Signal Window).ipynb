```python
from collections import deque

n = int(input("Enter no of signal readings: "))
arr = list(map(int, input("Enter the signal values: ").split()))
k = int(input("k: "))

max_dq = deque()
min_dq = deque()
left = 0
max_length = 0
start_position = 1  # HERE I  took base 1  

for right in range(n):
    # Maintain decreasing deque for maximums
    while max_dq and arr[max_dq[-1]] <= arr[right]:
        max_dq.pop()
    max_dq.append(right)

    # Maintain increasing deque for minimums
    while min_dq and arr[min_dq[-1]] >= arr[right]:
        min_dq.pop()
    min_dq.append(right)

    # now if Shrink window if difference between max and min exceeds k
    while arr[max_dq[0]] - arr[min_dq[0]] > k:
        if max_dq[0] == left:
            max_dq.popleft()
        if min_dq[0] == left:
            min_dq.popleft()
        left += 1

    # lets Update longest window length and start position
    current_length = right - left + 1
    if current_length > max_length:
        max_length = current_length
        start_position = left + 1  

print(max_length, start_position)
```
